# 400 MeV Booster Space Charge Tune Shift

From Schindl, Space Charge

Eq. 43

$$ \Delta Q_{x,y} = - \frac{r_0}{\pi} \left ( \frac{q^2}{A} \right )
    \frac{N}{\beta^2 \gamma^3}
    \frac{F_{x,y} G_{x,y} }{B_f}
    \frac{1}{E_{x,y} (1 + \sqrt{\frac{E_{y,x}Q_{x,y}}{E_{x,y}Q_{y,x}} } ) }
$$

* $F_{x,y}$ form factor derived from Laslett's image coefficients for incoherent tune shifts.
    * $F_x$ = $b^2/(a(a+b))$, $F_y$ = $b/(a+b)$
* $G_{x,y}$ is form factor depending on distribution. $G=1$ for uniform, $E_{x,y}$ contains 100%, $G=2$ fpr Gaussian, $E_{x,y}$ is $4 \sigma^2/\beta$.
* $B_f$ is longitudinal bunching factor mean/peak density. $B_f = 1$ for unbunched, $B_f < 1$ for bunched.
* $q$ is the charge state.
* $A$ is the ion mass.


```python
from numpy import pi, sqrt
from scipy.constants import eV, c, m_p, epsilon_0

mp_mev = 1.0e-6 * m_p * c**2/eV
```

```python
mp_mev
```

```python
r_0 = eV/(4 * pi * epsilon_0 * mp_mev * 1.0e6)
```

```python
r_0
```

```python
NBooster = 5.5e12
Active_Buckets = 81
Nbunch = NBooster/Active_Buckets
print('Protons/bunch: ', Nbunch, (1.0e-10*Nbunch),'e10')
```

# Booster Parameters at 400 MeV

```python
L = 474.202752 # Booster orbit length
emit_x = 2.077e-6 # geometric
emit_y = 1.732e-6 # geometric
emit_z = 0.00161 # ct-dp/p
beta_z = 879.2 # sig-ct/sig-dpop

alpha_x = -0.03374203811363915
beta_x = 33.432491467233
alpha_y = 0.011403474422954477
beta_y = 5.2633267186119905

disp_x = 3.071236819185704 # /(dpop)
disp_px = 0.003057679536025426 # /(dpop)
# taken from turn 50 whwn acceleration has started after capture
ke = 400.1545611624435 # MeV
energy = ke + mp_mev
gamma = energy/mp_mev
beta_gamma = sqrt((gamma - 1)*(gamma+1))
beta = beta_gamma/gamma

sig_dpop = sqrt(emit_z/beta_z)
sig_x = sqrt(emit_x * beta_x + (disp_x * sig_dpop)**2)
sig_y = sqrt(emit_y * beta_y)
sig_cdt = sqrt(emit_z * beta_z)

# Base tunes
Q_x = 6.71
Q_y = 6.81
```

```python
print('gamma: ', gamma)
print('beta: ', beta)
print('sig-x: ', sig_x)
print('sig-y: ', sig_y)
print('sig-cdt: ',  sig_cdt)
print('sig-dpop: ', sig_dpop)
print('Q_x: ', Q_x)
print('Q_y: ', Q_y)
```

```python
# For Laslett coefficients
# a = <extent x> = sqrt(<epsilon beta>) = sqrt(epsilon>) sqrt(<beta>)
# <beta> = L/(2 pi nu)


a = sqrt(emit_x * L/(2*pi*Q_x))
b = sqrt(emit_y * L/(2*pi*Q_y))
F_x = b**2/(a * (a + b))
F_y = b/(a + b)
print('a: ', a)
print('b: ', b)
print("Laslett coefficients: ")
print("    F_x: ", F_x)
print("    F_y: ", F_y)
```

```python
# G_{x,y} is 2 for Gaussian
G_x = 2
G_y = 2
```

```python
# calculate B_f as mean/peak for Gaussian. Use +- 3 sigma for mean
g_mean = 1/6
g_peak = 1/sqrt(2*pi)
B_f = g_mean/g_peak
print("bunch factor B_f: ", B_f)
```

```python
dQ_x = (r_0/pi) * (Nbunch/(beta**2 * gamma**3)) * (F_x*G_x/B_f) * \
1/(4*emit_x * (1 + sqrt( (emit_y*Q_x)/(emit_x*Q_y) ) ) )
```
